# HessianFactor

A `HessianFactor` in GTSAM represents a linear factor in terms of a Hessian (information) matrix and a gradient vector. It is useful for directly constructing factors from second-order information, and is a key component in Gaussian (linear) factor graphs.

## Mathematical Form

A Hessian factor encodes a quadratic error of the form:

$$
\frac{1}{2} x^T H x - g^T x + c
$$

where $H$ is the Hessian (information) matrix, $g$ is the gradient vector, and $c$ is a constant. The error is:

$$
\text{error}(x) = \frac{1}{2} x^T H x - g^T x + c
$$

```python
# Import required libraries
import gtsam
import numpy as np
```

## Creating a HessianFactor

We create a simple `HessianFactor` for a single variable. The Hessian matrix $H$, gradient vector $g$, and constant $c$ are specified as numpy arrays.

```python
# Define variable key
key = 0

# Define Hessian block, gradient, and constant
H = np.array([[2.0, 0.0], [0.0, 2.0]])  # 2x2 Hessian
b = np.array([4.0, 6.0])                # Gradient vector
c = 1.0                                 # Constant

# Create the HessianFactor
factor = gtsam.HessianFactor(key, H, b, c)

# Add to a GaussianFactorGraph
graph = gtsam.GaussianFactorGraph()
graph.add(factor)
print(f"Graph size after adding factor: {graph.size()}")
```

## Accessing Factor Properties

We can inspect the keys, print the factor, and examine its Hessian block and gradient.

```python
# Print the factor
graph.at(0).print()

# Get keys
print(f"Keys: {list(factor.keys())}")

# Get Hessian block and gradient
block = factor.info(key)
grad = factor.linearTerm(key)
print(f"Hessian block for key {key}:\n{block}")
print(f"Gradient for key {key}: {grad}")
```

## Evaluating Error

We can compute the error for a given assignment of variable values using a `gtsam.Values` object.

```python
# Create Values and insert assignment
values = gtsam.Values()
values.insert(key, np.array([1.0, 2.0]))

# Compute error
err = factor.error(values)
print(f"Error at given values: {err}")
```
