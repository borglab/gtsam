## Example (numpy fallback)

As an example, form a small linear system where Gaussian factors correspond to rows of A and entries of b.

```python
A = np.array([[1.0, 0.0], [0.0, 1.0]])
b = np.array([0.5, -0.5])
x = np.linalg.solve(A, b)
print('Trivial solve (numpy):', x)
x
```
