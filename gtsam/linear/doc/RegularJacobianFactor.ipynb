## Example (numpy fallback)

Form a small Jacobian corresponding to a regular factor and solve least-squares.

```python
J = np.array([[1.0,0.0],[0.0,1.0]])
b = np.array([0.5,-0.5])
x, *_ = np.linalg.lstsq(J, b, rcond=None)
print('LS solution (numpy):', x)
x
```
