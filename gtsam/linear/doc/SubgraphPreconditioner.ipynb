## Conceptual example

Show building a simple preconditioner from a submatrix.

```python
A = np.array([[4.0,1.0],[1.0,3.0]])
sub = A
M_inv = np.eye(2)/np.diag(sub)
print('Conceptual preconditioner (numpy):', M_inv)
M_inv
```
