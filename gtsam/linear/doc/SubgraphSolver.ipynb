## Example (numpy)

Solve a submatrix as a demonstration of a subgraph solve.

```python
A = np.array([[3.0,0.0],[0.0,2.0]])
b = np.array([1.0, -1.0])
x = np.linalg.solve(A, b)
print('Sub-solve (numpy):', x)
x
```
