`GaussianEliminationTree` represents the elimination structure for Gaussians. This notebook documents the concept and provides guarded examples for Python bindings.

```python
import gtsam
import numpy as np
import graphviz
```

## Creating a GaussianEliminationTree

The Python bindings may or may not expose the class; we handle both cases below.

```python
try:
    getree = gtsam.GaussianEliminationTree()
except Exception as e:
    getree = None
    print('GaussianEliminationTree not available in Python bindings:', e)
getree
```

## Example (numpy fallback)

Demonstrate elimination with a small linear system solved by numpy.

```python
A = np.array([[4.0, -1.0], [-1.0, 3.0]])
b = np.array([1.0, 2.0])
x = np.linalg.solve(A, b)
print('Solve (numpy):', x)
x
```
