## Creating a GaussianFactorGraph

Guarded creation example for Python bindings.

## Example (numpy fallback)

Form and solve a small linear system representing a Gaussian factor graph.

```python
A = np.array([[3.0, -1.0], [-1.0, 2.0]])
b = np.array([1.0, 0.0])
x = np.linalg.solve(A, b)
print('Solve (numpy):', x)
x
```
