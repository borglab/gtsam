## Using iterative utilities

Guarded examples for Python bindings followed by a numpy demonstration of an iterative method.

```python
# Simple Jacobi iteration (numpy)
A = np.array([[4.0, -1.0], [-1.0, 3.0]])
b = np.array([15.0, 10.0])
x = np.zeros(2)
for _ in range(10):
    x = (b - (A - np.diag(np.diag(A))) @ x) / np.diag(A)
print('Jacobi iterate (numpy):', x)
x
```
