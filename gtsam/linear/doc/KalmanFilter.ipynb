## Simple Kalman step (numpy)

Perform a single predict-update cycle for a 1D constant model.

```python
# State: x, variance P
x = 0.0
P = 1.0
# Model: x' = x, process noise Q
Q = 0.1
# Predict
x = x
P = P + Q
# Measurement z = 0.5, R
z = 0.5
R = 0.2
K = P / (P + R)
x = x + K * (z - x)
P = (1 - K) * P
print('Post-update state:', x, 'P:', P)
(x, P)
```
