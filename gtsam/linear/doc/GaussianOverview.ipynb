## GaussianFactor and GaussianFactorGraph

`GaussianFactor` objects contribute linear rows to a system; `GaussianFactorGraph` collects them and can solve the resulting linear system. Below we show guarded construction and a small numpy example.

```python
# Guarded construction examples
try:
    gf = gtsam.GaussianFactor()
    gfg = gtsam.GaussianFactorGraph()
except Exception as e:
    gf = None
    gfg = None
    print('GaussianFactor/GaussianFactorGraph not available in Python bindings:', e)

# Numpy example: form and solve a small system
A = np.array([[3.0, -1.0], [-1.0, 2.0]])
b = np.array([1.0, 0.0])
x = np.linalg.solve(A, b)
print('Solve (numpy):', x)
x
```

## GaussianBayesNet and GaussianConditional

A `GaussianBayesNet` is a sequence of `GaussianConditional`s (conditionals). We attempt to construct them if available and otherwise show how a conditional relates to block matrices.

```python
try:
    gbn = gtsam.GaussianBayesNet()
    gc = gtsam.GaussianConditional()
except Exception as e:
    gbn = None
    gc = None
    print('GaussianBayesNet/GaussianConditional not available in Python bindings:', e)

# Numpy illustration: solve a small conditional block
A = np.array([[2.0, -1.0], [-1.0, 2.0]])
b = np.array([1.0, 0.0])
x = np.linalg.solve(A, b)
print('Example conditional solve (numpy):', x)
x
```

## GaussianBayesTree and GaussianEliminationTree

Bayes trees and elimination trees capture elimination structure for efficient inference. If bindings exist we construct them; otherwise we show a small elimination example using numpy solves.

```python
try:
    gbt = gtsam.GaussianBayesTree()
    getree = gtsam.GaussianEliminationTree()
except Exception as e:
    gbt = None
    getree = None
    print('GaussianBayesTree/GaussianEliminationTree not available in Python bindings:', e)

# Numpy elimination demo
A = np.array([[4.0, -1.0], [-1.0, 3.0]])
b = np.array([1.0, 2.0])
x = np.linalg.solve(A, b)
print('Elimination example (numpy):', x)
x
```

## GaussianDensity and GaussianJunctionTree

`GaussianDensity` stores densities (unnormalized); `GaussianJunctionTree` supports junction-tree algorithms. Below we show a pdf evaluation and a linear solve example.

```python
try:
    gd = gtsam.GaussianDensity()
    gjt = gtsam.GaussianJunctionTree()
except Exception as e:
    gd = None
    gjt = None
    print('GaussianDensity/GaussianJunctionTree not available in Python bindings:', e)

# Numpy gaussian pdf example
x = np.array([0.1, -0.2])
mu = np.zeros(2)
Sigma = np.eye(2)
det = np.linalg.det(Sigma)
norm = 1.0 / (2*np.pi*np.sqrt(det))
val = norm * np.exp(-0.5 * (x-mu).T @ np.linalg.inv(Sigma) @ (x-mu))
print('Gaussian pdf (numpy):', val)
val
```

## GaussianISAM (incremental)

`GaussianISAM` is an incremental solver. We try to construct it and otherwise show a simple incremental idea with numpy.

```python
try:
    gisam = gtsam.GaussianISAM()
except Exception as e:
    gisam = None
    print('GaussianISAM not available in Python bindings:', e)

# Numpy incremental illustration (single step)
A = np.array([[2.0, 0.0],[0.0,2.0]])
b = np.array([1.0, 1.0])
x = np.linalg.solve(A, b)
print('Incremental example (numpy):', x)
x
```
