## Running the Power Method

Guarded example for bindings; otherwise run a numpy implementation below.

```python
# Numpy power iteration
A = np.array([[2.0, 0.5],[0.5,1.0]])
v = np.random.randn(2)
for _ in range(20):
    v = A @ v
    v = v / np.linalg.norm(v)
print('Estimated eigenvector (numpy):', v)
v
```
