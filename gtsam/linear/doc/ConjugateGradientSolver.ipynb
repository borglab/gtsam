## Example (numpy fallback)

Solve Ax=b using numpy to demonstrate the expected result of a CG-like solver.

```python
A = np.array([[4.0, 1.0], [1.0, 3.0]])
b = np.array([1.0, 2.0])
x = np.linalg.solve(A, b)
print('Solution (numpy):', x)
x
```
