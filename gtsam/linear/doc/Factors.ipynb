## JacobianFactor and BinaryJacobianFactor

Jacobian factors represent linear rows; binary variants connect two variable blocks. We try to construct them and show a numpy least-squares example.

```python
try:
    jf = gtsam.JacobianFactor()
    bjf = gtsam.BinaryJacobianFactor()
except Exception as e:
    jf = None
    bjf = None
    print('JacobianFactor/BinaryJacobianFactor not available in Python bindings:', e)

# LS example (numpy)
J = np.array([[1.0, 2.0],[3.0,4.0]])
b = np.array([1.0,2.0])
x, residuals, rank, s = np.linalg.lstsq(J, b, rcond=None)
print('Least-squares (numpy):', x)
x
```

## HessianFactor and RegularHessianFactor

Hessian factors work in information form. We try to construct them and show a small MAP solve via numpy.

```python
try:
    hf = gtsam.HessianFactor()
    rhf = gtsam.RegularHessianFactor()
except Exception as e:
    hf = None
    rhf = None
    print('HessianFactor/RegularHessianFactor not available in Python bindings:', e)

# MAP solve (numpy)
H = np.array([[4.0,0.0],[0.0,2.0]])
g = np.array([-2.0,-2.0])
x = np.linalg.solve(H, -g)
print('MAP estimate (numpy):', x)
x
```

## RegularJacobianFactor

Regular Jacobian factors have block regular structure; we include a small numpy example for least-squares.

```python
try:
    rjf = gtsam.RegularJacobianFactor()
except Exception as e:
    rjf = None
    print('RegularJacobianFactor not available in Python bindings:', e)

# Example (numpy)
J = np.eye(2)
b = np.array([0.5,-0.5])
x, *_ = np.linalg.lstsq(J, b, rcond=None)
print('LS (numpy):', x)
x
```
