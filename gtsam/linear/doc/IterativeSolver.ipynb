## Example (numpy fallback)

Use numpy to run a simple iterative solve (e.g. CG via numpy.linalg for small systems).

```python
A = np.array([[4.0, 1.0], [1.0, 3.0]])
b = np.array([1.0, 2.0])
x = np.linalg.solve(A, b)
print('Direct solution (numpy):', x)
x
```
