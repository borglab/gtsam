## Example losses

Show simple Huber and Quadratic loss evaluations (numpy fallback).

```python
def huber(r, k=1.0):
    return np.where(np.abs(r) <= k, 0.5*r**2, k*(np.abs(r)-0.5*k))
r = np.array([0.1, 2.0, -0.5])
print('Huber loss (numpy):', huber(r))
```
