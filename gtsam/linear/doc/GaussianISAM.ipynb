## Example (numpy fallback)

Show a simple incremental update concept using numpy linear solves.

```python
A = np.array([[2.0, 0.0], [0.0, 2.0]])
b = np.array([1.0, 1.0])
x = np.linalg.solve(A, b)
print('Example solve (numpy):', x)
x
```
