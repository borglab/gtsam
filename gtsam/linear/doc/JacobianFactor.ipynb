## Example (numpy fallback)

Form a small Jacobian and solve the least-squares problem with numpy.

```python
J = np.array([[1.0, 0.0], [0.0, 1.0]])
b = np.array([0.5, -0.5])
x, residuals, rank, s = np.linalg.lstsq(J, b, rcond=None)
print('Least-squares (numpy):', x)
x
```
