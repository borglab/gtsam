## Example (numpy fallback)

Form a simple Hessian and linear term and solve for the MAP estimate.

```python
H = np.array([[4.0, 0.0], [0.0, 2.0]])
g = np.array([-2.0, -2.0])
x = np.linalg.solve(H, -g)
print('MAP estimate (numpy):', x)
x
```
