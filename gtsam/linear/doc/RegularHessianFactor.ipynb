## Example (numpy fallback)

Construct a simple regular Hessian and solve for the MAP estimate.

```python
H = np.array([[3.0, 0.0],[0.0,2.0]])
g = np.array([-1.0,-1.0])
x = np.linalg.solve(H, -g)
print('MAP (numpy):', x)
x
```
