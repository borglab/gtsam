```python
import gtsam
import numpy as np
import graphviz
```

## Creating PCGSolver

Guarded construction for Python bindings.

```python
try:
    pcg = gtsam.PCGSolver()
except Exception as e:
    pcg = None
    print('PCGSolver not available in Python bindings:', e)
pcg
```

## Example (numpy fallback)

Use numpy for a direct solve as a reference for PCG result.

```python
A = np.array([[6.0, 2.0],[2.0,5.0]])
b = np.array([1.0, 2.0])
x = np.linalg.solve(A, b)
print('Direct solution (numpy):', x)
x
```
