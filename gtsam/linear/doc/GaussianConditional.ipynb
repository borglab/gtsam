## Example (numpy fallback)

Show a small conditional-like operation using numpy block matrices.

```python
A = np.array([[1.0, 0.5], [0.5, 1.0]])
b = np.array([1.0, 0.0])
x = np.linalg.solve(A, b)
print('Example solve (numpy):', x)
x
```
