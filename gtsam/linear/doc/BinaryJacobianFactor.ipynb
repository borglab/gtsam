## Example (numpy fallback)

Demonstrate an equivalent small example using numpy to show Jacobian concepts.

```python
J = np.array([[1.0, 2.0], [3.0, 4.0]])
b = np.array([1.0, 2.0])
x, residuals, rank, s = np.linalg.lstsq(J, b, rcond=None)
print('Least-squares solution (numpy):', x)
x
```
