## Example (numpy fallback)

Compute eigenvalues of a small dense matrix with numpy as a substitute for sparse routines.

```python
A = np.array([[2.0, 0.5],[0.5,1.0]])
w, v = np.linalg.eigh(A)
print('Eigenvalues (numpy):', w)
w
```
