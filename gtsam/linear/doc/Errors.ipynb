## Example: compute residuals and norms

Demonstrate basic error computations using numpy as a fallback.

```python
A = np.array([[2.0, -1.0], [-1.0, 2.0]])
x_true = np.array([1.0, 0.0])
b = A @ x_true
x_est = np.array([0.9, 0.1])
res = b - A @ x_est
print('Residual:', res)
print('L2 norm of residual:', np.linalg.norm(res))
res
```
