## ConjugateGradientSolver and PCGSolver

Try to create solver objects and otherwise solve with numpy (direct solve as reference).

```python
try:
    cgs = gtsam.ConjugateGradientSolver()
    pcg = gtsam.PCGSolver()
except Exception as e:
    cgs = None
    pcg = None
    print('ConjugateGradientSolver/PCGSolver not available in Python bindings:', e)

# Reference solve (numpy)
A = np.array([[6.0,2.0],[2.0,5.0]])
b = np.array([1.0,2.0])
x = np.linalg.solve(A, b)
print('Reference solve (numpy):', x)
x
```

## IterativeSolver and iterative utilities

Show a simple iterative method (Jacobi) as a demonstration.

```python
try:
    its = gtsam.IterativeSolver()
except Exception as e:
    its = None
    print('IterativeSolver not available in Python bindings:', e)

# Jacobi iteration (numpy)
A = np.array([[4.0,-1.0],[-1.0,3.0]])
b = np.array([15.0,10.0])
x = np.zeros(2)
for _ in range(10):
    x = (b - (A - np.diag(np.diag(A))) @ x) / np.diag(A)
print('Jacobi iterate (numpy):', x)
x
```

## Preconditioners and Subgraph-based methods

Show a diagonal preconditioner and conceptual subgraph preconditioning/solving examples.

```python
try:
    pre = gtsam.Preconditioner()
    sgb = gtsam.SubgraphBuilder()
    sgs = gtsam.SubgraphSolver()
    sgp = gtsam.SubgraphPreconditioner()
except Exception as e:
    pre = None
    sgb = sgs = sgp = None
    print('Preconditioner/Subgraph* bindings not available:', e)

# Diagonal preconditioner (numpy)
A = np.array([[10.0,1.0],[1.0,5.0]])
M_inv = np.diag(1.0/np.diag(A))
print('Diagonal preconditioner (M^-1):', M_inv)
M_inv
```

## PowerMethod and SparseEigen

Compute dominant eigenvector with numpy power iteration as a fallback for sparse eigen helpers.

```python
# Numpy power iteration
A = np.array([[2.0,0.5],[0.5,1.0]])
v = np.random.randn(2)
for _ in range(20):
    v = A @ v
    v = v / np.linalg.norm(v)
print('Estimated eigenvector (numpy):', v)
v
```
