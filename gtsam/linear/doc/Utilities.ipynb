## NoiseModel and LossFunctions

Try constructing common noise models and evaluate simple loss functions (Huber) as examples.

```python
try:
    nm = gtsam.noiseModel.Isotropic.Sigma(1, 0.1)
except Exception as e:
    nm = None
    print('noiseModel bindings not available:', e)

# Huber loss (numpy)
def huber(r, k=1.0):
    return np.where(np.abs(r) <= k, 0.5*r**2, k*(np.abs(r)-0.5*k))
r = np.array([0.1, 2.0, -0.5])
print('Huber loss (numpy):', huber(r))
```

Show sampling and simple vector-values storage examples.

## Scatter and Errors, linearExceptions

Show COO-style arrays and a basic residual error computation; demonstrate catching an example exception.

```python
try:
    scat = gtsam.Scatter()
except Exception as e:
    scat = None
    print('Scatter bindings not available:', e)

rows = [0,1,2]
cols = [0,2,1]
vals = [1.0,2.0,3.0]
print('COO-like arrays (numpy):', rows, cols, vals)
# Residual error demo
A = np.array([[2.0,-1.0],[-1.0,2.0]])
x_true = np.array([1.0,0.0])
b = A @ x_true
x_est = np.array([0.9,0.1])
res = b - A @ x_est
print('Residual:', res, 'L2 norm:', np.linalg.norm(res))
# Example catch
try:
    raise RuntimeError('example linear exception')
except Exception as e:
    print('Caught exception:', e)
```

## KalmanFilter

A short numpy-based Kalman filter step is shown as a reference for the linear Kalman utilities.

```python
# 1D Kalman update example
x = 0.0
P = 1.0
Q = 0.1
# Predict
x = x
P = P + Q
z = 0.5
R = 0.2
K = P / (P + R)
x = x + K * (z - x)
P = (1 - K) * P
print('Post-update state:', x, 'P:', P)
```
