## Example (numpy fallback)

Compute a simple Gaussian pdf value for a vector using numpy.

```python
x = np.array([0.1, -0.2])
mu = np.zeros(2)
Sigma = np.eye(2)
det = np.linalg.det(Sigma)
norm = 1.0 / (2*np.pi*np.sqrt(det))
val = norm * np.exp(-0.5 * (x-mu).T @ np.linalg.inv(Sigma) @ (x-mu))
print('Gaussian pdf (numpy):', val)
val
```
