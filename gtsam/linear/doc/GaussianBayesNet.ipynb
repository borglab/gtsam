## Example (numpy fallback)

Show a small chain model and the factorization idea using numpy matrices.

```python
A = np.array([[2.0, -1.0], [-1.0, 2.0]])
b = np.array([1.0, 0.0])
x = np.linalg.solve(A, b)
print('Example solution (numpy):', x)
x
```
