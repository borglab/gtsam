## Parameter groups

- `pMin` and `pMax` bound the Burer–Monteiro factorization rank.
- `alpha` scales the new column used to escape a certified saddle direction.
- `eta` is the PSD certificate tolerance.
- `verificationMethod` selects sparse Spectra verification or dense Eigen verification.
- `maxSpectraIters`, `numLanczosVectors`, and `spectraTol` tune the sparse eigensolver fallback.
- `getAlmParams()` and `setAlmParams()` copy the inner augmented-Lagrangian parameters across the Python boundary.

```python
alm = gtsam.AugmentedLagrangianParams()
alm.maxIterations = 80
alm.absoluteViolationTolerance = 1e-8

params = gtsam.RiemannianStaircaseParams()
params.pMin = 2
params.pMax = 5
params.alpha = 1e-2
params.eta = 1e-3
params.verificationMethod = gtsam.RiemannianStaircaseParams.VerificationMethod.Spectra
params.setAlmParams(alm)

copied_alm = params.getAlmParams()
print("rank range:", params.pMin, "to", params.pMax)
print("inner iterations:", copied_alm.maxIterations)
print("verification:", params.verificationMethod)
```

Rot2 matrix problems require `pMin >= 2`; Rot3 requires `pMin >= 3`. `DenseEigen` is useful for small validation problems, while `Spectra` avoids a dense cubic eigendecomposition and is the normal scalable choice.
