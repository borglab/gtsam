## Shared relaxation and recovery contract

Both specializations lift quadratic terms in $x$ to linear trace terms in $Z\succeq0$. They share the same public lifecycle:

- construction from `QcqpProblem`;
- `solve()` plus status, objective, and timing queries;
- functional `qcqpValues()` and `variableEVRs()` recovery queries;
- ordered QCQP keys and dimensions.

The monolithic formulation owns one PSD cone. The chordal formulation owns overlapping clique cones built from a symbolic Bayes tree. The formulation and solver tags are compile-time C++ details; Python uses the concrete class names below.

```python
classes = ["MosekMonolithicSDP", "MosekChordalSDP", "ChordalOrderingType"]
for name in classes:
    print(f"{name}: {'available' if hasattr(gtsam, name) else 'not in this build'}")
```

## Choosing a formulation

Use `MosekMonolithicSDP` as the simplest reference formulation and for small dense problems. Use `MosekChordalSDP` when factor-graph sparsity produces clique cones substantially smaller than the full lifted matrix. Both require GTSAM to be built with MOSEK and a valid license.

See [MosekMonolithicSDP](MosekMonolithicSDP.ipynb) and [MosekChordalSDP](MosekChordalSDP.ipynb) for complete Python solves and typed Pose recovery. For the solver-independent low-rank route, see [RiemannianStaircaseOptimizer](RiemannianStaircaseOptimizer.ipynb).
